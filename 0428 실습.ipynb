# 영문 번역

```python
from googletrans import Translator
translator = Translator()
result = translator.translate('안녕하세요', src='ko', dest='en')
print(result.text)  # "Hello"
```

```python
result = translator.translate('안녕하세요 저는 홍길동입니다. 만나서 반갑습니다.', src='ko', dest='en')
print(result.text)  # "Hello"
```
